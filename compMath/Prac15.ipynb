# Практическая работа 15: вейвлеты и другие методы работы с аудио

Вейвлеты это универсальный инструмент для анализа сигналов и изображений. По своей сути, они представляют собой небольшие осциллирующие функции, которые помогают разбивать данные на различные частотные составляющие, сохраняя при этом информацию о времени или пространстве.

В отличие от традиционных преобразований Фурье, которые преобразуют сигналы в чистые синусоиды (что приводит к потере временной локализации), вейвлеты позволяют проводить анализ с различным разрешением, увеличивая масштаб изображения и сохраняя общую картину. Это делает их невероятно полезными в таких областях, как сжатие изображений, определение границ и даже решение дифференциальных уравнений.

Их можно рассматривать как адаптируемые строительные блоки, которые могут выявить как глобальные закономерности, так и мелкие детали в наборе данных.

**Ключевое отличие** заключается в том, что преобразования Фурье разлагают сигналы на суммы синусоид (чистых колебаний бесконечной длительности), которые отлично подходят для анализа глобального частотного содержания, но теряют локализацию во времени или пространстве.

С другой стороны, вейвлеты используют функции, локализованные как по частоте, так и по времени, что означает, что они могут фиксировать резкие изменения и переходные процессы так, как это не удается синусоидам. Эти вейвлет-функции часто асимметричны и имеют компактную поддержку (они исчезают за пределами определенной области), что делает их особенно полезными для анализа сигналов с внезапными скачками или нерегулярностями.

Таким образом, вместо представления сигнала в виде суммы бесконечно длинных волн вейвлет-преобразования разбивают его на короткоживущие масштабируемые волнообразные структуры, обеспечивая более детальный и гибкий подход к анализу сигналов с различным разрешением. Это все равно что переключиться с размытого глобального изображения на масштабируемый микроскоп для получения характеристик сигнала!

Каждое семейство вейвлетов обладает уникальными характеристиками, адаптированными для различных приложений:

- **Вейвлет Хаара**: Простейший вейвлет, состоящий из ступенчатой функции с резкими переходами. Он отлично подходит для **быстрых и эффективных вычислений**, особенно при сжатии изображений (например, JPEG 2000). Однако он не является плавным, что ограничивает его использование в приложениях, требующих высокой детализации.

- **Вейвлеты Добеши**: Более сложные, чем у Haar, эти вейвлеты **плавные и компактно поддерживаемые**, что делает их идеальными для обработки сигналов и устранения помех. Они обеспечивают баланс между локализацией по времени и частоте, что помогает в таких приложениях, как медицинская визуализация и анализ речи.

- **Вейвлет Морле**: Синусоида с гауссовым окном, что делает его особенно эффективным для анализа сигналов с **локализованными частотными составляющими** (таких как ЭЭГ-волны головного мозга или сейсмические данные). Он часто используется в частотно-временном анализе, поскольку напоминает локализованное преобразование Фурье, но сохраняет гибкость вейвлет-анализа.

Каждое семейство служит своей цели: Haar - для повышения эффективности, Daubechies - для обеспечения плавности и компактности, а Morlet - для частотного анализа.

Важно помнить, что вейвлеты применяются далеко не только для работы с медиа! **Это обработка данных любой природы, будь то медицина, экономика, анализ рынка, IoT, астрономия и многое другое.**

**Вейвлеты**, хотя и остаются ценными, теперь считаются зрелым инструментом, представленным в 1980-х годах (Маллат, Добеши). С тех пор **вычислительная математика и обработка сигналов** значительно развились, особенно за последние **20 лет**.

Вот краткий обзор **современных достижений**, которые расширили или заменили вейвлеты во многих приложениях:

## 🔥 Основные достижения со времен вейвлетов

### 1. **Разреженные представления и сжатый анализ / Sparse Representations & Compressed Sensing (2006 – настоящее время)**

* **Ключевая идея**: Сигналы могут быть восстановлены по *очень небольшому количеству выборок (samples)*, если они разрежены в некотором базисе (часто это вейвлеты или DCT  дискретное косинусное преобразование).
* **Влияние**: Революционная медицинская визуализация (например, МРТ-сканирование в 4-8 раз быстрее), радар, беспроводная связь.
* **Алгоритмы**: Лассо, поиск ортогонального соответствия (OMP - Orthogonal Matching Pursuit), поиск базиса (Basis Pursuit).

> Изобретатели: Кандес, Ромберг, Тао, Донохо

---

### 2. **Неотрицательная матричная факторизация (NMF, Nonnegative Matrix Factorization)**

* **Ключевая идея**: Разложить спектрограммы на аддитивные части — это удобно для интерпретации.
* **Варианты использования**: Разделение источников (например, отделение голоса от музыки), моделирование речи, биоакустика.
* Часто предпочтительнее вейвлетов для **разреженной и интерпретируемой** декомпозиции.

> Впервые сформулированно: Lee & Seung (1999)

---

### 3. **Вариационные методы и шумоподавление на основе PDE (дифференциальных уравнений в часных производных)**

* **Шумоподавление с полной вариацией (TV - Total Varuation)**: лучше сохраняет границы, чем вейвлет-пороговое значение.
* **Приложения**: Медицинская визуализация, обработка видео, шумоподавление без размытия.

> Связано с: Модель Рудина–Ошера–Фатеми (1992-2000–е)

---

### 4. **Графовая обработка сигналов (GSP)**

* **Сигналы на графах/сетях** — например, массивы датчиков, сигналы мозга, социальные сети.
* **Теперь существуют вейвлет-преобразования для графов**.
* **Приложения**: ЭЭГ, Интернет вещей, социальные данные, прогнозирование трафика.

> Последние годы: 2013 – настоящее время

---

### 5. **Извлечение признаков на основе глубокого обучения**

* **Спектрограмма + CNNs** (свёрточные NN) и **модели необработанных сигналов** превосходят традиционные преобразования во многих областях.
* Модели *автоматически изучают лучшие базовые функции*, превосходя по производительности семейства фиксированных вейвлетов.
* **Примеры архитектур**:

  * **WaveNet**, **Conv-TasNet**, **wav2vec 2.0**, **Whisper**
  * Использование 1D / 2D свертки вместо банков вейвлетов

> Заменяет: разработанные вручную базовые функции (например, вейвлеты)
> Появился: с 2014 года
---

### 6. **Нейронное сжатие и квантование**

* Вдохновлен вейвлетами, но использует **изученные преобразования и квантователи** (learned transforms and quantizers).
* **Примеры**:

  * EnCodec (Meta)
  * SoundStream
  * VQ-VAE / Residual VQ
* **Приложения**: Синтез речи, аудио с низким битрейтом, генеративные модели
---

### 7. **Трансформаторы (transformers) сигналов**

* Трансформаторы взяли верх над НЛП и теперь также используются для аудио / речи/временных рядов.
* Такие модели, как **Audio Spectrogram Transformer (AST)** или **Perceiver**, превосходят классические методы, основанные на признаках.
* Заменяет: Необходимость в ручном масштабировании или разработке [MFCC](https://habr.com/ru/articles/140828/).

---

### 8. **Модели распространения сигналов (Diffusion Models)**

* Применяются для синтеза звука, устранения шума и даже для решения обратных задач, таких как сжатое распознавание.
* Используйте изученные модели шума вместо фиксированных пороговых значений, как в вейвлетах.

> Например, синтез распространения звука (Google), рисование изображений / аудио

# Вначале попробуем применить вейвлеты для того, чтобы убрать шум

```python
import pywt
import numpy as np

def wavelet_denoise(signal, wavelet='db8', level=4, threshold_scale=0.5):
    coeffs = pywt.wavedec(signal, wavelet, level=level)
    sigma = np.median(np.abs(coeffs[-1])) / 0.6745
    uthresh = sigma * np.sqrt(2 * np.log(len(signal))) * threshold_scale
    coeffs_thresh = [pywt.threshold(c, value=uthresh, mode='soft') for c in coeffs]
    return pywt.waverec(coeffs_thresh, wavelet)

denoised_audio = wavelet_denoise(audio)
```

```python
import io
import soundfile as sf

buffer = io.BytesIO()
denoised_audio_np = denoised_audio.astype(np.float32)
sf.write(buffer, denoised_audio_np, samplerate=sr, format='WAV')
buffer.seek(0)
Audio(buffer.read())
```

# Теперь попробуем сжать аудио (нейронное сжатие и квантование)
Для этого можно попробовать модель encodec от запрещённой в России Meta Inc. Или любую другую на Ваш выбор. В данном случае токенизация поможет нам сжать аудио.

```python
import torchaudio
import torch
from encodec import EncodecModel
from encodec.utils import convert_audio

# Load pretrained EnCodec
model = EncodecModel.encodec_model_24khz()
model.set_target_bandwidth(6.0)  # Set compression rate

# Convert your numpy audio to torch tensor
waveform = torch.from_numpy(denoised_audio).float().unsqueeze(0).unsqueeze(0)
waveform = convert_audio(waveform, sr, model.sample_rate, model.channels)

# Encode
with torch.no_grad():
    encoded_frames = model.encode(waveform)
    codes = [frame[0] for frame in encoded_frames]  # RVQ levels × tokens
```

```python
decoded_audio = model.decode([(c, None) for c in codes])
```

```python
import torchaudio
import soundfile as sf

# Remove batch and channel dimensions: [1, 1, T] → [T]
audio_np = decoded_audio.squeeze().detach().cpu().numpy()

# Save to file (e.g., WAV)
sf.write('decoded.wav', audio_np, samplerate=model.sample_rate)
```

# --- Вейвлеты ---

Вейвлеты - это математические функции, которые разделяют сигналы на различные частотные составляющие, локализованные как во времени, так и по частоте. Это делает их идеальными для анализа нестационарных сигналов, таких как аудио, в отличие от преобразования Фурье, которое предоставляет только информацию о частоте.

Анализ с множественным разрешением (MRA) разбивает сигнал на приближенные значения (низкие частоты) и детализированные значения (высокие частоты) в различных масштабах. Например, вейвлет Хаара представляет собой простую пошаговую функцию, которая иллюстрирует эту концепцию.

```python
import numpy as np
import matplotlib.pyplot as plt
import librosa
import pywt
from gtts import gTTS
from IPython.display import Audio, display
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

# --- Генерация аудио ---
Сгенерируем пример аудио на русском языке

```python
text = "Всем привет, я сижу в общаге, ем семечки. Сегодня из достижений я задеплоила третью лабу по вебу на гелиус."
tts = gTTS(text=text, lang='ru')
tts.save("russian_audio.mp3")

# Load the audio
audio, sr = librosa.load("russian_audio.mp3", sr=None)
print("Original Russian Audio:")
display(Audio(audio, rate=sr))

# Apply wavelet transform (Daubechies 4 wavelet)
wavelet = 'db4'
coeffs = pywt.wavedec(audio, wavelet, level=5)

# Visualize coefficients
plt.figure(figsize=(12, 8))
for i, c in enumerate(coeffs):
    plt.subplot(len(coeffs), 1, i + 1)
    plt.plot(c)
    plt.title(f'Level {i} Coefficients' if i == 0 else f'Detail Level {i}')
plt.tight_layout()
plt.show()
```

# --- Пример: Обнаружение переходов ---

Создадим синтетический сигнал с переходным процессом

```python
t = np.linspace(0, 1, 1000)
signal = np.sin(2 * np.pi * 7 * t) + 0.1 * np.random.randn(1000)
signal[200:220] += 5  # Add a transient spike
```

**Fourier Transform**

```python
fft_signal = np.fft.fft(signal)
freq = np.fft.fftfreq(len(t), t[1] - t[0])
plt.figure()
plt.plot(freq[:len(freq)//2], np.abs(fft_signal[:len(freq)//2]))
plt.title('Fourier Transform (Frequency Domain)')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Amplitude')
plt.show()
```

**Wavelet Transform**

```python
coeffs_transient = pywt.wavedec(signal, wavelet, level=5)
plt.figure(figsize=(12, 8))
for i, c in enumerate(coeffs_transient):
    plt.subplot(len(coeffs_transient), 1, i + 1)
    plt.plot(c)
    plt.title(f'Level {i} Coefficients' if i == 0 else f'Detail Level {i}')
plt.tight_layout()
plt.show()
```

**Примечание:** Вейвлеты локализуют переходный процесс во времени (например, коэффициенты детализации на уровне 1-2), в то время как БПФ распределяет его по частотам.

Example: Compression

```python
threshold = 0.1 * max(abs(coeffs[0]))
compressed_coeffs = [pywt.threshold(c, threshold, mode='soft') for c in coeffs]
compressed_audio = pywt.waverec(compressed_coeffs, wavelet)
print("Compressed Audio:")
display(Audio(compressed_audio[:len(audio)], rate=sr))
```

## 🔊 Wavelet Denoising

```python
def wavelet_denoise(signal, wavelet='db1', level=3):
    coeffs = pywt.wavedec(signal, wavelet, level=level)
    sigma = np.median(np.abs(coeffs[-1])) / 0.6745
    uthresh = sigma * np.sqrt(2 * np.log(len(signal)))
    coeffs = [pywt.threshold(c, value=uthresh, mode='soft') for c in coeffs]
    return pywt.waverec(coeffs, wavelet)
```

## 🧱 RESD Dataset

```python
class RESDDataset(Dataset):
    def __init__(self, root_dir, sr=16000, max_len=4.0, transform=None,
                 feature_type='scalogram', apply_denoise=False):
        self.root_dir = root_dir
        self.sr = sr
        self.max_len = int(sr * max_len)
        self.transform = transform
        self.feature_type = feature_type
        self.apply_denoise = apply_denoise
        self.file_paths = []
        self.labels = []
        self._collect_files()

    def _extract_label_from_path(path):
      """Try to infer emotion label from the path string."""
      path_lower = path.lower()
      for keyword in LABEL_MAP.keys():
          if keyword in path_lower:
              return LABEL_MAP[keyword]
      return None  # Unknown


    def _collect_files(self):
        for subdir, _, files in os.walk(self.root_dir):
            for file in files:
                if file.lower().endswith('.wav'):
                    full_path = os.path.join(subdir, file)
                    label = RESDDataset._extract_label_from_path(full_path)
                    if label is not None:
                        self.file_paths.append(full_path)
                        self.labels.append(label)


    def __len__(self):
        return len(self.file_paths)

    def __getitem__(self, idx):
        path = self.file_paths[idx]
        label = self.labels[idx]
        audio, _ = librosa.load(path, sr=self.sr)

        if self.apply_denoise:
            audio = wavelet_denoise(audio)

        if len(audio) < self.max_len:
            audio = np.pad(audio, (0, self.max_len - len(audio)))
        else:
            audio = audio[:self.max_len]

        if self.feature_type == 'scalogram':
            features = self._get_scalogram(audio)
        else:
            raise ValueError("Use 'scalogram' as feature_type")

        return features, label

    def _get_scalogram(self, signal):
        scales = np.arange(1, 64)
        coeffs, _ = pywt.cwt(signal, scales, 'morl')
        power = np.abs(coeffs)
        power = power[:64, :256]  # Crop/pad to fixed size
        return torch.tensor(power).float()
```

## 🧠 CNN Model

Классическая CNN, которая будет обучаться "в лоб", даже если результаты становятся хуже.

```python
class EmotionCNN(nn.Module):
    def __init__(self, num_classes=7):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 16, 3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.conv3 = nn.Conv2d(32, 64, 3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.dropout = nn.Dropout(0.3)
        # Adjusted input size for fc1 based on calculated flattened dimensions (64 * 7 * 32)
        self.fc1 = nn.Linear(64 * 7 * 32, 128)
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = x.unsqueeze(1)  # (B, 1, H, W) where H=63, W=256
        x = self.pool(F.relu(self.conv1(x))) # (B, 16, 31, 128)
        x = self.pool(F.relu(self.conv2(x))) # (B, 32, 15, 64)
        x = self.pool(F.relu(self.conv3(x))) # (B, 64, 7, 32)
        x = x.view(x.size(0), -1) # Flatten to (B, 64 * 7 * 32)
        x = self.dropout(F.relu(self.fc1(x)))
        return self.fc2(x)
```

Альтернативно можно создать `EarlyStopping` CNN, которая будет завершать обучение, если всё идёт не по плану.

```python
import pytorch_lightning as pl
import torch.nn.functional as F

class EmotionClassifier(pl.LightningModule):
    def __init__(self, num_classes=7, lr=1e-3):
        super().__init__()
        self.save_hyperparameters()
        self.cnn = EmotionCNN(num_classes)

    def forward(self, x):
        return self.cnn(x)

    def training_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = F.cross_entropy(logits, y)
        acc = (logits.argmax(dim=1) == y).float().mean()
        self.log("train_loss", loss)
        self.log("train_acc", acc)
        return loss

    def validation_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = F.cross_entropy(logits, y)
        acc = (logits.argmax(dim=1) == y).float().mean()
        self.log("val_loss", loss, prog_bar=True)
        self.log("val_acc", acc, prog_bar=True)

    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=self.hparams.lr)
```

## 🏋️ Train Model

Обучение для классической CNN **(занимает 2 часа, не делайте это пожалуйста на паре).**

```python
model = EmotionCNN().to(torch.device('cuda' if torch.cuda.is_available() else 'cpu'))
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

train_losses, val_losses, val_accuracies = [], [], []
print(datetime.datetime.now())
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

for epoch in range(10):
    model.train()
    running_loss = 0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    train_losses.append(running_loss / len(train_loader))

    # Validation
    model.eval()
    val_loss, correct, total = 0, 0, 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            _, preds = torch.max(outputs, 1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)

    val_losses.append(val_loss / len(val_loader))
    val_accuracies.append(correct / total)

    print(f"{datetime.datetime.now()} Epoch {epoch+1}: Train Loss={train_losses[-1]:.4f}, Val Acc={val_accuracies[-1]:.4f}")
```
